This is just to combine the output files, borrowing the position on LAB computer. 

```python
from pathlib import Path
import pandas as pd

run_names = [
    "iterations_20260616_200626_seq2seq",
    "iterations_20260617_150300_seq2seq",
    # add more here
]

run_model = "seq2seq"

all_tables = []
iteration_offset = 0

for run_name in run_names:
    root_dir = Path("output") / run_name

    run_tables = []

    for pred_path in root_dir.glob("iteration_*/**/predictions.csv"):
        df = pd.read_csv(pred_path)

        # Shift iteration numbers so they are unique across runs
        df["iteration"] = df["iteration"] + iteration_offset

        run_tables.append(df)

    if run_tables:
        run_df = pd.concat(run_tables, ignore_index=True)

        max_iter = run_df["iteration"].max()
        iteration_offset = max_iter + 1

        all_tables.append(run_df)

combined_df = pd.concat(all_tables, ignore_index=True)
combined_df["model_type"] = run_model

combined_df.to_csv(
    f"combined_predictions_all_runs_{run_model}.csv",
    index=False
)

print(
    f"Combined {len(all_tables)} runs into "
    f"{len(combined_df)} rows."
)
```
